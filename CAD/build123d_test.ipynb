```python
import build123d
import ocp_vscode

ocp_vscode.find_and_set_port()
port = ocp_vscode.get_port()
if port is None:
    raise RuntimeError("No OCP CAD Viewer is running. Open it from the VS Code sidebar, then rerun this cell.")

ocp_vscode.show_clear()
```

```python
length, width, thickness = 80.0, 60.0, 10.0
center_hole_dia = 22.0

with build123d.BuildPart() as ex2:
    build123d.Box(length, width, thickness)
    build123d.Cylinder(radius=center_hole_dia / 2, height=thickness, mode=build123d.Mode.SUBTRACT)
```

```python
radius, height = 80, 10
hole_radius = 5
bore_depth = height / 2

with build123d.BuildPart() as ex13:
    cylinder_part = build123d.Cylinder(radius=radius, height=height)
    with build123d.Locations(ex13.faces().sort_by(build123d.Axis.Z)[-1]) as top_plane:
        with build123d.PolarLocations(radius=radius - 2 * hole_radius - 1, count=4) as counter_sink_hole_locations:
            sink_holes = build123d.CounterSinkHole(radius=hole_radius, counter_sink_radius=1.5 * hole_radius)
        with build123d.PolarLocations(radius=radius - 2 * hole_radius - 1, count=4, start_angle=45, angular_range=360) as counter_bore_hole_locations:
            bore_holes = build123d.CounterBoreHole(radius=hole_radius, counter_bore_radius=1.5 * hole_radius, counter_bore_depth=bore_depth)
```
